```python
import numpy as np
import matplotlib.pyplot as plt
```

## Part 1

```python
def lorentz_normalized(x, alpha_ratio):
    """
    Calculates the normalized Lorentz line shape
    x = (nu - nu_0) / alpha_0
    alpha_ratio = alpha / alpha_0
    """
    numerator = alpha_ratio
    denominator = x**2 + alpha_ratio**2
    return numerator / denominator

# Set up the data
x = np.linspace(-4, 4, 400) # Normalized frequency axis
alpha_ratio_0 = 1.0         # Surface (T0 = 20C)
alpha_ratio_T = 0.791       # Atmosphere (T = 1C)

y_0 = lorentz_normalized(x, alpha_ratio_0)
y_T = lorentz_normalized(x, alpha_ratio_T)

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, y_0, color='red', 
         label=r'T$_0$ = 20$\degree$C (Surface, $\alpha=\alpha_0$)')
plt.plot(x, y_T, color='blue', 
         label=r'T = 1$\degree$C (Atmosphere, $\alpha=0.791\alpha_0$)')

# Labels and title
plt.xlabel(r'Frequency difference from center $(\nu - \nu_0) / \alpha_0$')
plt.ylabel(r'Normalized Line Shape ($\Phi_L \cdot \pi \alpha_0$)')
plt.title('Lorentz Line Shape at Different Temperatures')
plt.legend()
plt.grid(True)
```
